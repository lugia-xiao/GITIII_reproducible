# mouse

```python
data=torch.load("./banksy/mouse.pth",weights_only=False)
print(data)
```

```python
adata=data.loc["scaled_gaussian_pc20_nc0.20_r0.08","adata"]
adata
```

```python
cell_type="L2/3 IT"
resolution="scaled_gaussian_pc20_nc0.20_r0.10"

adata_type=adata[adata.obs["subclass"]==cell_type]

unique_labels=np.unique(adata_type.obs["labels_"+resolution])
print(unique_labels)
mapping={}
cnt=1
for labeli in unique_labels:
    mapping[labeli]="s"+str(cnt)+"-"+cell_type
    cnt=cnt+1
print(mapping)
new_labels=[]
for labeli in adata_type.obs["labels_"+resolution]:
    new_labels.append(mapping[labeli])
adata_type.obs["banksy_subgroup"]=np.array(new_labels)

sc.pl.scatter(
        adata_type,
        x='xcoord',#'Spatial x',
        y='ycoord',#'Spatial y',
        color="banksy_subgroup",
        title="resolution="+resolution[-4:]
    )
```
