```python
import pandas as pd
import numpy as np
import scanpy as sc
import anndata as ad
from pathlib import Path

# --- Config ---
# If you already have a list of gene names, keep it as `genes` and skip inference below.
# genes = [...]

# Map filenames -> Batch (segmentation method) labels
files_to_batch = {
    "BIDCell.csv": "BIDCell",
    "JSTA.csv": "JSTA",
    "v.csv": "Voronoi",     # user said v.csv; full name is "voronoi"
}

data_dir = Path("../data")  # adjust if needed

# --- Helpers ---
NON_GENE_COLS = {"centerx", "centery", "subclass"}

def load_one_csv_to_adata(csv_path: Path, batch_label: str, genes_list=None) -> ad.AnnData:
    df = pd.read_csv(csv_path)
    # Normalize column names (strip spaces)
    df.columns = df.columns.str.strip()

    # Determine gene columns
    if genes_list is None:
        # Infer: all columns except known metadata
        genes_inferred = [c for c in df.columns if c not in NON_GENE_COLS]
    else:
        genes_inferred = list(genes_list)

    # Basic checks
    missing = [g for g in genes_inferred if g not in df.columns]
    if missing:
        raise ValueError(
            f"{csv_path.name}: the following genes are missing in the file: {missing[:10]} "
            f"{'(showing first 10)' if len(missing) > 10 else ''}"
        )

    # Expression matrix
    X = df[genes_inferred].to_numpy(dtype=float)

    # obs (cell-level metadata)
    obs = pd.DataFrame(index=df.index)
    # Ensure these columns exist; if not, create with NaNs
    for col in ["centerx", "centery", "subclass"]:
        if col in df.columns:
            obs[col] = df[col].values
        else:
            obs[col] = np.nan
    obs["Batch"] = batch_label
    # Make subclass categorical for convenience
    obs["subclass"] = pd.Categorical(obs["subclass"])

    # var (gene-level metadata)
    var = pd.DataFrame(index=genes_inferred)

    adata = ad.AnnData(X=X, obs=obs, var=var)
    return adata

# --- Load each file into an AnnData then concatenate ---
adatas = []
for fname, batch in files_to_batch.items():
    path = data_dir / fname
    if not path.exists():
        raise FileNotFoundError(f"Expected file not found: {path.resolve()}")
    # Pass `genes` if you already have it defined; otherwise omit it and we’ll infer
    try:
        _ = genes  # check if user provided
        adata_i = load_one_csv_to_adata(path, batch_label=batch, genes_list=genes)
    except NameError:
        adata_i = load_one_csv_to_adata(path, batch_label=batch, genes_list=None)
    adatas.append(adata_i)

# Outer-join genes across batches (in case of small mismatches), fill missing with 0
adata = ad.concat(adatas, join="outer", label=None, axis=0, merge="first")
adata.X = np.nan_to_num(adata.X, copy=False)

# Optional: enforce a consistent gene order (alphabetical or as provided)
# If you supplied `genes`, keep that order; otherwise sort alphabetically
try:
    _ = genes
    # Reindex var to supplied order; add any extra genes (if outer join added some)
    extra = [g for g in adata.var_names if g not in genes]
    adata = adata[:, list(genes) + extra]
except NameError:
    adata = adata[:, sorted(adata.var_names)]

# Quick sanity checks / summary
print(adata)
print("Batches:", adata.obs["Batch"].value_counts().to_dict())
print("Subclasses (top 10):")
print(adata.obs["subclass"].value_counts().head(10))

cell_type_corrected=[]
for i in adata.obs["subclass"]:
    if i=='Firoblast':
        cell_type_corrected.append('Fibroblast')
    else:
        cell_type_corrected.append(i)

adata.obs['subclass']=np.array(cell_type_corrected)

# Optional: save to disk
adata.write("combined.h5ad")
```

```python
import numpy as np
import scanpy as sc
import pandas as pd

# Nice-looking figures
sc.set_figure_params(figsize=(5, 4), dpi=120)

# Targets; handle the "Fibroblast" vs "Fibroblast" spelling by normalizing to "Fibroblast"
targets = ["Macrophage", "Fibroblast", "SCGB2A2+ Malignant"]
normalize_map = {"Fibroblast": "Fibroblast"}

# Make a normalized subclass column for selection
subclass_norm = adata.obs["subclass"].astype(str).map(lambda x: normalize_map.get(x, x))
adata.obs["_subclass_norm"] = pd.Categorical(subclass_norm)

# Ensure Batch is categorical
adata.obs["Batch"] = adata.obs["Batch"].astype("category")

def umap_for_subset(label: str):
    ad_sub = adata[adata.obs["_subclass_norm"] == label].copy()
    n_cells = ad_sub.n_obs
    if n_cells < 5:
        print(f"Skip '{label}': only {n_cells} cells.")
        return

    # Recompute on the subset (data already normalized/QC'd)
    n_pcs = min(50, max(5, int(np.ceil(min(ad_sub.n_vars, n_cells - 1, 50)))))
    n_neighbors = int(max(5, min(15, n_cells - 1)))

    sc.pp.pca(ad_sub, n_comps=n_pcs, svd_solver="arpack", random_state=0)
    sc.pp.neighbors(ad_sub, n_neighbors=n_neighbors, n_pcs=n_pcs, metric="euclidean")
    sc.tl.umap(ad_sub, min_dist=0.4, spread=1.0, random_state=0)

    # Show UMAP colored by Batch (no saving)
    sc.pl.umap(
        ad_sub,
        color="Batch",
        size=8,
        alpha=0.9,
        legend_loc="right margin",
        frameon=False,
        title=f"{label}: UMAP colored by Batch",
        show=True,
    )

for lbl in targets:
    umap_for_subset(lbl)

# Clean up helper column if you want:
# adata.obs.drop(columns=["_subclass_norm"], inplace=True)
```
