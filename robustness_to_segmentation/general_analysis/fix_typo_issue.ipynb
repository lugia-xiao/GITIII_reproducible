```python
import torch
import numpy as np
import pandas as pd
import copy
```

```python
method="BIDCell"
file_path="../"+method+"/influence_tensor/edges_sample01.pth"
data=torch.load(file_path,weights_only=False)
```

```python
for i in range(len(data['cell_type_name'])):
    for j in range(len(data['cell_type_name'][0])):
        if data['cell_type_name'][i][j]=='Firoblast':
            data['cell_type_name'][i][j]='Fibroblast'
torch.save(data,file_path)
```

```python
method="BIDCell"
file_path="../"+method+"/data/processed/cell_types.pth"
data=torch.load(file_path,weights_only=False)
data=["Fibroblast" if x == "Firoblast" else x for x in data]
torch.save(data,file_path)
```

```python
file_path="../"+method+"/data/processed/sample01_TypeExp.npz"

# Load the .npz into a mutable dict
data_npz = np.load(file_path)
data = {k: copy.deepcopy(v) for k, v in data_npz.items()}

# Rename the key
if "Firoblast" in data:
    data["Fibroblast"] = data.pop("Firoblast")

# Save back to .npz if needed
np.savez(file_path, **data)
```

```python
file_path="../"+method+"/data/processed/sample01.csv"
df=pd.read_csv(file_path, index_col=0)
new_type=[]
for i in df.loc[:,"subclass"]:
    if i=="Firoblast":
        new_type.append("Fibroblast")
    else:
        new_type.append(i)
df["subclass"]=np.array(new_type)
df.to_csv(file_path)
```
